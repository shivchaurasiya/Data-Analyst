```python
df.rename(columns={
    "Student Name": "student_Name",
    "EMAIL Address": "email",
    "Age (yrs)": "age",
    "Gender": "gender",
    "City/Location": "city",
    "Department Name": "department",
    "Course": "course",
    "Marks %": "marks",
    "Annual Salary": "salary",
    "Phone Number": "phone",
    "Admission Date": "admission_date",
    "Attendance %": "attendance",
    "Student_ID": "student_id"
}, inplace=True)

df.columns
```

```python
missing_count = df.isna().sum()
missing_percent = (df.isna().mean() *100).round(2)

missing_report = pd.DataFrame({
    'missing_count': missing_count,
    'missing_percent': missing_percent
}).sort_values(by='missing_percent', ascending=False)
missing_report
```

```python
import numpy as np

missing_tokens = ["", "NA", "null", "Not Available", "unknown"]

df = df.replace(missing_tokens, np.nan)

print("Total missing values:", df.isna().sum().sum())

df.isna().sum().sort_values(ascending=False)
```

```python
# Convert numric columns to numeric data type
df['age']=pd.to_numeric(df['age'],errors='coerce')
df['marks']=pd.to_numeric(df['marks'],errors='coerce')
df['attendance']=pd.to_numeric(df['attendance'],errors='coerce')
df['salary']=pd.to_numeric(df['salary'],errors='coerce')
```

```python
df.columns
```

```python
for col in ["Gender ", "city", "department", "course"]:
    df[col] = df[col].fillna("Unknown")

df["age"] = df["age"].fillna(df["age"].median())

df["salary"] = df["salary"].fillna(df["salary"].median())

df["marks"] = df["marks"].fillna(df["marks"].median())

df["attendance"] = df["attendance"].fillna(df["attendance"].median())
```

```python
df.describe()
```

```python
df['city'].unique()
```

```python
# Remove extra spaces from column names
df.columns = df.columns.str.strip()

text_columns = ["student_Name", "email", "Gender", "city", "department", "course"]

# Strip whitespace from text columns
for col in text_columns:
    df[col] = df[col].astype("string").str.strip()

# Convert selected categorical columns to lowercase
for col in ["Gender", "city", "department", "course"]:
    df[col] = df[col].str.lower()

# Display first 11 rows
display(df[["student_Name", "Gender", "city", "department", "course"]].head(11))
```

```python
df.drop_duplicates(inplace=True)
print("Duplicates:",df.duplicated().sum())
```

```python
print("New Shape:",df.shape)
print("Duplicates remaining:",df.duplicated().sum())
```

```python
numeric_cols = ['age','marks','salary','attendance']
for col in numeric_cols:
  df[col] = pd.to_numeric(df[col],errors='coerce')

print(df[numeric_cols].describe())
```

```python
#Triner soln : Convert inpossible values to NAn
df.loc[~df['age'].between(17,35),'age'] = np.nan
df.loc[~df['marks'].between(0,100),'marks'] = np.nan
df.loc[df['salary'] <=0,'salary'] = np.nan
df.loc[~df['attendance'].between(0,100),'attendance'] = np.nan

# Refill values after invalid records were converted to nan
df['age'] = df['age'].fillna(df['age'].median())
df['marks'] = df['marks'].fillna(df['marks'].median())
df['salary'] = df['salary'].fillna(df['salary'].median())
df['attendance'] = df['attendance'].fillna(df['attendance'].median())
```

```python
# Clean email
df["email"] = df["email"].astype("string").str.strip().str.lower()

# Find invalid emails
invalid_emails = ~df["email"].str.match(
    r"^[^\s@]+@[^\s@]+\.[^\s@]+$", na=False
)

print("Invalid emails found:", invalid_emails.sum())

# Replace invalid emails with NaN
df.loc[invalid_emails, "email"] = np.nan


# Clean phone numbers
df["phone"] = (
    df["phone"]
    .astype("string")
    .str.replace(r"\D", "", regex=True)
)

# Find invalid phone numbers
invalid_phone = ~df["phone"].str.match(
    r"^[6-9]\d{9}$", na=False
)

print("Invalid phone numbers found:", invalid_phone.sum())

# Replace invalid phone numbers with NaN
df.loc[invalid_phone, "phone"] = np.nan
```

```python
df["admission_date"] = pd.to_datetime(df["admission_date"], errors="coerce",dayfirst=True)
print("Missing/invalid admission dates:", df["admission_date"].isna().sum())
print(df["admission_date"].head(10))
```
